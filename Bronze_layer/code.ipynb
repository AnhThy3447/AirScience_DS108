# merging file

```python
import pandas as pd
import os
```

```python
os.getcwd()
```

```python
merge_file = []
for file in os.listdir():
    if file.endswith('.csv'):
        df = pd.read_csv(file)
        print(file)
        print(df.shape)
        # print(df.columns)
        merge_file.append(df)
merge_file = pd.concat(merge_file, ignore_index=True)
merge_file.to_csv('merged_file.csv', index=False)
```
